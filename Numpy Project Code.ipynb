```python
import pandas as pd
import numpy as np

df=pd.read_csv("Bank_transactions_data.csv")
transactions=df.to_numpy()

print(transactions.shape)
print(transactions.ndim)
print(transactions.size)
```

```python
#Task 2 — Understand the columns
# 0  → TransactionID
# 1  → AccountID
# 2  → TransactionAmount
# 3  → TransactionDate
# 4  → TransactionType
# 5  → Location
# 6  → DeviceID
# 7  → IP Address
# 8  → MerchantID
# 9  → Channel
# 10 → CustomerAge
# 11 → CustomerOccupation
# 12 → TransactionDuration
# 13 → LoginAttempts
# 14 → AccountBalance
#Before extracting anything, let's see the column names from the DataFrame:
print(df.columns)
print()
print(transactions[0])
```

```python
#Task 3 — Extract the important numerical columns
# Let's start the actual analysis.
# Write code to extract these four columns:
# TransactionAmount
# CustomerAge
# LoginAttempts
# AccountBalance

amounts=transactions[:,2]
ages=transactions[:,10]
balances=transactions[:,14]
login_attempts=transactions[:,13]

print(amounts.shape)
print(ages.shape)
print(login_attempts.shape)
print(balances.shape)
```

```python
#Task 4 — Transaction Amount Analysis
# Now let's use the NumPy aggregation methods you've already learned.
# Using only amounts, calculate:
# Total transaction amount
# Average transaction amount
# Minimum transaction amount
# Maximum transaction amount
# Standard deviation

total_amount=np.sum(amounts)
average_amount=np.mean(amounts)
minimum_amount=np.min(amounts)
maximum_amount=np.max(amounts)
std_amount=np.std(amounts)
print("Total amount:",total_amount)
print("Average amount:",average_amount)
print("Minimum amount:",minimum_amount)
print("Maximum amount:",maximum_amount)
print("Std of amount:",std_amount)

#One useful observation: the standard deviation (~₹293) is almost as large as the mean (~₹298). 
#That tells us transaction amounts have considerable variation.
```

```python
#Task 5 — Find unusually large transactions
# Now let's use Boolean indexing, one of the most important NumPy skills for Data Analyst/Data Engineer work.
# Find all transaction amounts that are greater than ₹1,000.

large_transactions=amounts[amounts>1000]
print(large_transactions)

#Then calculate the percentage of transactions that are above ₹1,000.
count=len(large_transactions)
percentage=(count/len(amounts))*100
print(percentage)

#We found that:
# 3.576% of all transactions are above ₹1,000.
# Since there are 50,000 transactions, that's about 1,788 transactions above ₹1,000.
```

```python
#Task 6 — Customer Age Analysis
# Let's now analyze the ages array.
# Calculate:
# Average customer age
# Minimum age
# Maximum age
# Number of customers above age 60

average_customer_age=np.mean(ages)
minimum_age=np.min(ages)
maximum_age=np.max(ages)
number_of_customers_above_age_60=len(ages[ages>60])
percentage_of_customers_above_age_60=(number_of_customers_above_age_60/len(ages))*100

print("Average customer age:",average_customer_age)
print("Minimum age:",minimum_age)
print("Maximum age:",maximum_age)
print("Number of customers above age 60:",number_of_customers_above_age_60)
print("Percentage above 60:",percentage_of_customers_above_age_60)
print()

#How to verify how results?
# Total observations
print(len(ages))

# Mean verification
print(np.sum(ages) / len(ages))

# Minimum verification
print(np.min(ages))

# Maximum verification
print(np.max(ages))

# Count verification — two different approaches
print(len(ages[ages > 60]))
print(np.sum(ages > 60))
```

```python
#Task 7 — Login Attempts
# Calculate these four metrics using login_attempts:
# 1. Minimum login attempts
# 2. Maximum login attempts
# 3. Average login attempts
# 4. Number of transactions where login attempts > 2

minimum_login_attempts=np.min(login_attempts)
maximum_login_attempts=np.max(login_attempts)
avg_login_attempts=np.mean(login_attempts)
transactions_with_morethan2_attempts=len(login_attempts[login_attempts > 2])
print("Minimum login attempts:",minimum_login_attempts)
print("Maximum login attempts:",maximum_login_attempts)
print("Average login attempts:",avg_login_attempts)
print("Transactions with more than 2 attempts:",transactions_with_morethan2_attempts)
print()
percentage = (1924 / 50000) * 100
print("percentage:",percentage)
print()
print("Validate the count")
count = np.sum(login_attempts > 2)
print(count) #both are same 1924
```

```python
#Task 8 — Account Balance Analysis
# Now let's analyze your balances array.
# Calculate:
# Average account balance
# Minimum account balance
# Maximum account balance
# Number of accounts with balance greater than ₹50,000

average_balance = np.mean(balances)
minimum_balance = np.min(balances)
maximum_balance = np.max(balances)

accounts_above_50000 = len(balances[balances > 50000])

print("Average account balance:", average_balance)
print("Minimum account balance:", minimum_balance)
print("Maximum account balance:", maximum_balance)
print("Accounts with balance above ₹50,000:", accounts_above_50000)

#Observation :
# your average of ₹5,122.74 lies between the minimum and maximum, 
# which is another basic sanity check.
```

```python
#Task 9 — Identify High-Value Transactions
# We already know:
# amounts → transaction amount
# ages → customer age
# login_attempts → login attempts
# balances → account balance
# Let's create a simple business-risk filter using two conditions:
# Find transactions where TransactionAmount > ₹1,000 AND LoginAttempts > 2

high_value_transaction_with_high_logins=transactions[
                                        (amounts>1000) & (login_attempts>2)
                                        ] 
print(high_value_transaction_with_high_logins)
print(high_value_transaction_with_high_logins.shape)
print("Number of transaction with amount>1000 and login attempts>2")
count_high_risk = len(high_value_transaction_with_high_logins)
print("Number of matching transactions:", count_high_risk)
```

```python
#Task 10 — Transaction Type Analysis
# Our dataset has a TransactionType column at index 4.
# First extract it:
transaction_type = transactions[:, 4]
print(transaction_type.shape)
print()

number_of_debit_transactions=len(transaction_type[transaction_type=="Debit"])
number_of_credit_transactions=len(transaction_type[transaction_type=="Credit"])
percentage_of_debit_transactions=(number_of_debit_transactions/len(transaction_type))*100

print("Number of Debit transactions:",number_of_debit_transactions)
print("Number of Credit transactions:",number_of_credit_transactions)
print("Percentage of Debit transactions:",percentage_of_debit_transactions) #yes, 77.494% Debit means roughly 3 out of every 4 transactions in this dataset are Debit.
print()

#One useful validation: You can verify that Debit + Credit = total:
print(number_of_debit_transactions + number_of_credit_transactions) 
```

```python
#Task 11 — Location Analysis
#Now we'll analyze the Location column.

#Step 1: Extract the Location column
locations=transactions[:,5]
print(locations.shape)
print()

# Step 2: Your task
# Find:
# What are the unique locations?
unique_locations=np.unique(locations)
print(unique_locations)
print()

# How many unique locations are present?
unique_locations_count=len(unique_locations)
print(unique_locations_count)
print()

#Step 3 — Count transactions by location
#How many transactions happened in each location?
location_name,location_counts=np.unique(locations,return_counts=True)
#return_counts=True means "Give me the unique values and also tell me how many times each value appears."

for location,count in zip(location_name,location_counts):
    print(location,":",count)
print()

#Step 4 Find the location with the most transactions
max_transactions_count=np.max(location_counts)
print(max_transactions_count)

max_transactions_locations=np.argmax(location_counts) #np.argmax(location_counts) → gives the index/position where that highest count occurs
print(location_name[max_transactions_locations])
```

```python
#Task 12 - Transaction Channel Analysis
#Step 1: Extract the column
channels=transactions[:,9]
print(channels.shape)
print()

#Step 2: Find the unique channels
unique_channels=np.unique(channels)
print(unique_channels)
print(len(unique_channels)) 
print()

#Step 3 — Count transactions by channel
channel_name,channel_count=np.unique(channels,return_counts=True)

for channel,count in zip(channel_name,channel_count):
    print(channel,":",count)
print()

#Step 4 — Find the most-used channel
most_used_channel=np.argmax(channel_count) #finds the position of most used channel
print(channel_name[most_used_channel]) #gives channel which is most used channel
print(channel_count[most_used_channel]) #gives the count of most used channel
```

```python
#Task 13 — Customer Occupation Analysis
#Step 1 — Extract the column
occupations = transactions[:, 11]
print(occupations.shape)
print()

#Step 2 — Find unique occupations
unique_occupations = np.unique(occupations)
print(unique_occupations)
print(len(unique_occupations))
print()

#Step 3 — Count transactions by occupation
occupation_names, occupation_counts = np.unique(
    occupations,
    return_counts=True
)

for occupation, count in zip(occupation_names, occupation_counts):
    print(occupation, ":", count)
print()

#Step 4 — Find the occupation with the most transactions
most_followed_occupation=np.argmax(occupation_counts)
print(occupation_names[most_followed_occupation])
print(occupation_counts[most_followed_occupation])
print()

#Step 5 — Find the percentage
student_percentage = (occupation_counts[most_followed_occupation] / len(occupations)) * 100
print(student_percentage)
```

```python
#Task 14: Transaction Duration Analysis
# Extract Transaction Duration
transaction_duration = transactions[:, 12]

# Basic statistics
average_duration = np.mean(transaction_duration)
minimum_duration = np.min(transaction_duration)
maximum_duration = np.max(transaction_duration)

# Transactions taking more than 100 seconds
long_transactions = transaction_duration[transaction_duration > 100]
count_long_transactions = len(long_transactions)

# Percentage
percentage_long_transactions = (
    count_long_transactions / len(transaction_duration)
) * 100

print("Average transaction duration:", average_duration)
print("Minimum transaction duration:", minimum_duration)
print("Maximum transaction duration:", maximum_duration)
print("Transactions above 100 seconds:", count_long_transactions)
print("Percentage above 100 seconds:", percentage_long_transactions)
```

```python
#Task 15 — Broadcasting + Business Calculation
# Now we'll use the NumPy broadcasting concept we already learned.
# Suppose the bank charges a 2% processing fee on every transaction.

amounts=transactions[:,2].astype(float)
print("Original Amount:",amounts[:10])
processing_fee=amounts*0.02
final_amount=amounts+processing_fee
print("2% fee:",np.round(processing_fee[:10],2))
print("Final amount after 2% fee:",np.round(final_amount[:10],2))
```

```python
#Task 16 — np.where() Business Categorization
# Create a transaction category based on amount:
# Amount > 1000 → "HIGH"
# Otherwise → "NORMAL"
# Use np.where():
transaction_category = np.where(
    amounts > 1000,
    "HIGH",
    "NORMAL"
)

print(transaction_category[:20])
high_count = np.sum(transaction_category == "HIGH")
print(high_count)
print()
high_percentage = (high_count / len(amounts)) * 100
print(high_percentage)
```

```python
#Task 17 — Missing Values & Data Cleaning
amounts = transactions[:, 2].astype(float)
ages = transactions[:, 10].astype(float)
balances = transactions[:, 14].astype(float)
login_attempts = transactions[:, 13].astype(float)
transaction_duration = transactions[:, 12].astype(float)

print("Missing amounts:", np.sum(np.isnan(amounts)))
print("Missing ages:", np.sum(np.isnan(ages)))
print("Missing balances:", np.sum(np.isnan(balances)))
print("Missing login attempts:", np.sum(np.isnan(login_attempts)))
print("Missing transaction duration:", np.sum(np.isnan(transaction_duration)))
```
